```python
class Employee:
    company = "Apple"
    def __init__(self, name, salary):
        self.name = name
        self.salary = salary

    def printDetails(self):
        print(f"this is company of {self.name} he owns the {self.company} with a huge salary of ${self.salary}")

e = Employee("Upendra", 10000000)
e.printDetails()
    
```
