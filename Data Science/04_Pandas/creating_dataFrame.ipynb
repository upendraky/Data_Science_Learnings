```python
import sqlite3
import pandas as pd

# Connect database
conn = sqlite3.connect("mydb.sqlite")

# Create table
conn.execute("""
CREATE TABLE IF NOT EXISTS users (
    id INTEGER PRIMARY KEY,
    name TEXT,
    lang TEXT
)
""")

# Insert data
conn.execute("INSERT INTO users (name, lang) VALUES ('Harry', 'Python')")
conn.execute("INSERT INTO users (name, lang) VALUES ('Jack', 'JavaScript')")

conn.commit()

# Read table
df = pd.read_sql("SELECT * FROM users", conn)

print(df)

# Close connection
conn.close()
```
