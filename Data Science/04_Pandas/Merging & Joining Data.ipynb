```python
import numpy as np 
import pandas as pd
```

```python
employees = pd.DataFrame({
    "EmpID": [1, 2, 3],
    "Name": ["Alice", "Bob", "Charlie"],
    "DeptID": [10, 20, 30]
        })
departments = pd.DataFrame({
    "DeptID": [10, 20, 40],
    "DeptName": ["HR", "Engineering", "Marketing"]
        })
```

```python
employees
```

```python
departments
```

```python
pd.merge(departments, employees, on="DeptID")#inner join
```

```python
pd.merge(departments, employees, on="DeptID", how = "left")#left join
```

```python
pd.merge(departments, employees, on="DeptID", how = "right")#right join
```

```python
pd.merge(departments, employees, on="DeptID", how="outer")#outer join
```
