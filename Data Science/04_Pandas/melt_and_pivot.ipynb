```python
import pandas as pd
data = {
    "Name":["Bob","Ravi","Harry"],
    "Math":[55,53,56],
    "Eng":[34,43,65],
    "Chem":[44,76,34]
}
df = pd.DataFrame(data)
```

```python
print(df)
```

```python
df.melt(id_vars=["Name"],value_vars=["Math", "Eng", "Chem"], var_name="Subject", value_name="Score")
```

```python
import pandas as pd

data = [
    ["Bob", "Math", 55],
    ["Ravi", "Math", 53],
    ["Harry", "Math", 56],
    ["Bob", "Eng", 34],
    ["Ravi", "Eng", 43],
    ["Harry", "Eng", 65],
    ["Bob", "Chem", 44],
    ["Ravi", "Chem", 76],
    ["Harry", "Chem", 34]
]

df = pd.DataFrame(data, columns=["Name", "Subject", "Score"])

print(df)
```

```python
result = df.pivot(
    index="Name",
    columns="Subject",
    values="Score"
)

print(result)
```
