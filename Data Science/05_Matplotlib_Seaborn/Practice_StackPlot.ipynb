```python
import matplotlib.pyplot as plt
```

```python
days = [1, 2, 3, 4, 5, 6, 7] # Days of the week
studying = [3, 4, 3, 5, 4, 3, 4]
playing = [2, 2, 1, 1, 2, 3, 2]
watching_tv = [2, 1, 2, 2, 1, 1, 1]
sleeping = [5, 5, 6, 5, 6, 5, 5]
colors = ["gold","blue","pink","purple"]
labels = ["Studying","Playing","Watching_Tv","Sleeping"]
plt.stackplot(days,studying,playing,watching_tv,sleeping,colors=colors,labels=labels,alpha=0.8)
plt.legend(loc='upper right')
plt.title("Daily Routine Of Kids")
plt.xlabel("Days",color="Black")
plt.ylabel("Hours",color="Black")
plt.grid(True)
plt.show()
```
