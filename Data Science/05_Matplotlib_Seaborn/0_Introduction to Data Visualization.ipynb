```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
years = [1990, 1992, 1994, 1996, 1998, 2000, 2003, 2005, 2007, 2010]
kohli = [100, 0, 500, 800, 1100, 1300, 1500, 1800, 1900, 2100]
sehwag = [0, 300, 800, 1200, 1500, 1700, 1600, 1400, 1000, 0]
rohit = [400, 600, 1000, 1400, 1600, 1900, 1400, 1500, 1900, 1800]
```

```python
"""Assignment:
    Use:
    Create a plot comparing Kohli, Rohit Sharma, and Sehwag across 10 years of hypothetical runs.
    Labels
    Legends
    Colors
    Line styles
    One custom style"""
```

```python
plt.plot(years,sehwag,label="Virender Sehwag",color="blue")
plt.plot(years,rohit,label="Rohit Sharma ", color="pink")
plt.plot(years,kohli,label="Virat Kohli", color="green")
plt.xlabel("Years",color="purple")
plt.ylabel("Runs",color="olive")
plt.legend()
plt.grid(True)
plt.title("Comparing Kohli, Rohit Sharma, and Sehwag across 10 years")
plt.style.use("tableau-colorblind10")
plt.show()
```
