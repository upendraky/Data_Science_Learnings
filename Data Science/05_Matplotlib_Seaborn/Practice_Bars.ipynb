```python
import matplotlib.pyplot as plt
years = [1990, 1992, 1994, 1996, 1998, 2000, 2003, 2005, 2007, 2010]
runs = [500, 700, 1100, 1500, 1800, 1200, 1700, 1300, 900, 1500]
```

```python
import numpy as np
sachin = [500, 700, 1100, 1500, 1800, 1200, 1700, 1300, 900, 1500]
sehwag = [0, 200, 900, 1400, 1600, 1800, 1500, 1100, 800, 0]
kohli = [0, 0, 500, 800, 1100, 1300, 1500, 1800, 1900, 2100]
x = np.arange(len(years)) # index positions
width = 0.25
```

```python
plt.bar(x - width,sehwag,width = width,label="Virender",color="Blue")
plt.bar(x ,kohli,width = width,label="Virat",color="Gray")
plt.bar(x + width,sachin,width = width, label="Tendulkar",color="Purple")
plt.legend()
plt.xlabel("Runs Scored")
plt.ylabel("Year")
plt.xticks(x,years)
plt.tight_layout()
plt.show()
```

```python
import numpy as np
height = 0.25

plt.barh(x -  height,sehwag,height, label="Virender",color="Blue")
plt.barh(x ,kohli,height, label="Virat",color="Gray")
plt.barh(x + height,sachin, height, label="Tendulkar",color="Purple")

plt.yticks(x,years)
plt.xlabel("Runs Scored")
plt.ylabel("Year")
plt.legend()

plt.tight_layout()

plt.show()
```

```python
width = 0.25

plt.barh(x - width, sehwag, height=width, label="Virender")
plt.barh(x, kohli, height=width, label="Virat")
plt.barh(x + width, sachin, height=width, label="Tendulkar")
plt.yticks(x, years)
plt.xlabel("Runs Scored")
plt.ylabel("Year")
plt.legend()
plt.tight_layout()
plt.show()
```
